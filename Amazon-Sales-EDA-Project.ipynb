**PLOT 1 — Movies vs TV Shows (Donut)**

```python
import matplotlib.pyplot as plt

# 1. DEFINE THE MISSING COLORS HERE
BLUE = "#00A8E1"         # Vivid Prime Blue
PRIME_TEAL = "#15E6CD"   # Bright Teal
PRIME_DARK = "#0F172A"   # Dark slate/navy for text

print("\n[Plot 1] Content type distribution")
type_counts = df["Type"].value_counts()
 
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(
    type_counts,
    labels=type_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=[BLUE, PRIME_TEAL], # Now these variables exist!
    wedgeprops={"width": 0.45, "edgecolor": "white", "linewidth": 2},
    textprops={"fontsize": 13},
)

ax.set_title("Movies vs TV Shows", fontsize=15, fontweight="bold", color=PRIME_DARK, pad=20)

# 2. NOTE ON SAVING: 
# If 'save' is a custom function you wrote elsewhere, leave it as save(fig, ...).
# If not, use the standard matplotlib method below:
plt.savefig("01_content_type.png", bbox_inches="tight", dpi=300)
 
```

**PLOT 2 — Genre Distribution (Top 12)**

```python
import matplotlib.pyplot as plt

# Define the missing color variables (Amazon Prime branding colors)
PRIME_BLUE = "#00A8E1"  # Classic Prime Blue
PRIME_DARK = "#1A242F"  # Dark blue/gray for text and titles

print("[Plot 2] Genre distribution")
# 1. Get the top 12 genres
genre_counts = df["Genre(s)"].value_counts().head(12)

# 2. Reverse the order so the largest bar appears at the top
genre_counts_reversed = genre_counts.iloc[::-1]

fig, ax = plt.subplots(figsize=(10, 5))

# 3. Pass the cleanly reversed Series data to barh
bars = ax.barh(genre_counts_reversed.index, genre_counts_reversed.values, color=PRIME_BLUE)

ax.bar_label(bars, padding=4, fontsize=9, color=PRIME_DARK)
ax.set_xlabel("Count", fontsize=11)
ax.set_title("Top 12 Genres on Amazon Prime", fontsize=14, fontweight="bold", color=PRIME_DARK)
ax.spines[["top", "right"]].set_visible(False)

# Make sure your custom save function is defined, or use plt.savefig()
try:
    save(fig, "02_genre_distribution.png")
except NameError:
    # Fallback in case 'save' is also an undefined custom function
    plt.savefig("02_genre_distribution.png", bbox_inches="tight")
    plt.show()
```

**PLOT 4 — Content Released per Year**

```python
print("[Plot 4] Content per year")
year_counts = (
    df.dropna(subset=["Release Year"])
    .groupby(["Release Year", "Type"])
    .size()
    .reset_index(name="Count")
)
year_counts = year_counts[year_counts["Release Year"] >= 1990]

fig, ax = plt.subplots(figsize=(13, 5))
for ctype, grp in year_counts.groupby("Type"):
    ax.plot(grp["Release Year"], grp["Count"], marker="o", markersize=4, label=ctype)
ax.set_xlabel("Release Year", fontsize=11)
ax.set_ylabel("Number of Titles", fontsize=11)
ax.set_title("Amazon Prime Content Released per Year (≥ 1990)", fontsize=14, fontweight="bold", color=PRIME_DARK)
ax.legend()
ax.spines[["top", "right"]].set_visible(False)
save(fig, "04_content_per_year.png")
```

**PLOT 5 — Top 10 Countries of Origin**

```python
print("[Plot 5] Country of origin")
country_counts = df["Country of origin"].value_counts().head(10)

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(country_counts.index, country_counts.values,
              color=sns.color_palette("Set2", 10))
ax.bar_label(bars, padding=3, fontsize=9)
ax.set_xlabel("Country", fontsize=11)
ax.set_ylabel("Number of Titles", fontsize=11)
ax.set_title("Top 10 Countries of Origin", fontsize=14, fontweight="bold", color=PRIME_DARK)
ax.spines[["top", "right"]].set_visible(False)
plt.xticks(rotation=30, ha="right")
save(fig, "05_top_countries.png")
```

**PLOT 6 — Top 10 Languages**

```python
print("[Plot 6] Language distribution")
lang_counts = df["Language"].value_counts().head(10)

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(lang_counts.index[::-1], lang_counts.values[::-1], color=PRIME_TEAL)
ax.set_xlabel("Number of Titles", fontsize=11)
ax.set_title("Top 10 Languages on Amazon Prime", fontsize=14, fontweight="bold", color=PRIME_DARK)
ax.spines[["top", "right"]].set_visible(False)
save(fig, "06_language_distribution.png")
```

**PLOT 7 — Age Rating Distribution**

```python
print("[Plot 7] Age ratings")
age_counts = df["Age Rating"].value_counts()

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(age_counts.index, age_counts.values,
              color=sns.color_palette("muted", len(age_counts)))
ax.bar_label(bars, padding=3, fontsize=9)
ax.set_xlabel("Age Rating", fontsize=11)
ax.set_ylabel("Count", fontsize=11)
ax.set_title("Age Rating Distribution", fontsize=14, fontweight="bold", color=PRIME_DARK)
ax.spines[["top", "right"]].set_visible(False)
save(fig, "07_age_ratings.png")
```

**PLOT 8 — IMDb Score vs Votes (Scatter)**

```python
print("[Plot 8] Score vs Votes scatter")
plot_df = df.dropna(subset=["IMDb Score", "Number of IMDb Votes"])

fig, ax = plt.subplots(figsize=(9, 6))
for ctype, grp in plot_df.groupby("Type"):
    ax.scatter(np.log1p(grp["Number of IMDb Votes"]),
               grp["IMDb Score"],
               alpha=0.45, s=18, label=ctype)
ax.set_xlabel("log(1 + IMDb Votes)", fontsize=11)
ax.set_ylabel("IMDb Score", fontsize=11)
ax.set_title("IMDb Score vs Popularity (log votes)", fontsize=14, fontweight="bold", color=PRIME_DARK)
ax.legend()
ax.spines[["top", "right"]].set_visible(False)
save(fig, "08_score_vs_votes.png")
```

**PLOT 9 — Availability by Prime Region**

```python
print("[Plot 9] Availability by region")
region_avail = (
    df.groupby(["Prime region", "Is Available"])
    .size()
    .unstack(fill_value=0)
    .rename(columns={True: "Available", False: "Unavailable"})
)

fig, ax = plt.subplots(figsize=(9, 5))
region_avail.plot(kind="bar", stacked=True, ax=ax,
                  color=[PRIME_BLUE, "#cccccc"], edgecolor="white")
ax.set_xlabel("Prime Region", fontsize=11)
ax.set_ylabel("Number of Titles", fontsize=11)
ax.set_title("Content Availability by Prime Region", fontsize=14, fontweight="bold", color=PRIME_DARK)
ax.legend(title="Status")
plt.xticks(rotation=0)
ax.spines[["top", "right"]].set_visible(False)
save(fig, "09_availability_by_region.png")
```

**PLOT 11 — Correlation Heatmap**

```python
import matplotlib.pyplot as plt
import seaborn as sns

print("[Plot 11] Correlation heatmap")

# 1. Define your desired numerical columns
num_cols = ["Release Year", "IMDb Score", "Number of IMDb Votes", 
            "Duration_num", "Number of episodes"]

# 2. Filter out any columns that don't actually exist in df
valid_cols = [col for col in num_cols if col in df.columns]

# Optional: Alert you if something was missing
missing_cols = set(num_cols) - set(valid_cols)
if missing_cols:
    print(f"⚠️ Warning: Missing columns excluded from heatmap: {missing_cols}")

# 3. Only calculate correlation if we have columns to work with
if len(valid_cols) > 1:
    corr = df[valid_cols].corr()

    fig, ax = plt.subplots(figsize=(7, 6))
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
                center=0, ax=ax, linewidths=0.5, square=True)
    plt.show()
else:
    print("❌ Error: Not enough valid numerical columns found to create a heatmap.")
```
